```python
# Examine the distribution of unique categories in categorical column 
categorical_cols=df[['Amount of Credit Cards',
       'Nationality', 'Occupation', 'Fee Structure', 'Saving Accounts', 'Foreign Currency Account',
       'BRId','GenderId', 'IAId','Loyalty Classification','Properties Owned', 'Risk Weighting','Income Bands']]
for col in categorical_cols: 
    print(f"Value counts for {col}:")
    display(df[col].value_counts())
```

## UNIVARIATE ANALYSIS

```python
for i,predictor in enumerate(df[['Amount of Credit Cards',
       'Nationality', 'Occupation', 'Fee Structure', 'Saving Accounts', 'Foreign Currency Account',
       'BRId','GenderId', 'IAId','Loyalty Classification','Properties Owned', 'Risk Weighting','Income Bands']].columns):
    plt.figure(i)
    sns.countplot(data=df,x=predictor)
```

## BIVARIATE ANALYSIS

```python
for i,predictor in enumerate(df[['Amount of Credit Cards',
       'Nationality', 'Occupation', 'Fee Structure', 'Saving Accounts', 'Foreign Currency Account',
       'BRId','GenderId', 'IAId','Loyalty Classification','Properties Owned', 'Risk Weighting','Income Bands']].columns):
    plt.figure(i)
    sns.countplot(data=df,x=predictor,hue='Nationality')
```

```python
# Histplot of value counts for different occupations

for col in categorical_cols:
    if col=='Occupation':
        continue
    plt.figure(figsize=(8,4))
    sns.histplot(df[col])
    plt.title('Histogram of occupation count')
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
    
```
